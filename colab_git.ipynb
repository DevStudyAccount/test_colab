### 1. Colabから直接GitHubに保存する（基本）
最もシンプルな方法は、Colabのメニューから直接GitHubへコミットする方法です。

* **保存方法**: `ファイル` > `GitHub にコピーを保存` を選択します。
* **認証**: 初回実行時にGitHubへのアクセス権限を求められるので、連携を許可します。
* **リポジトリとブランチの選択**: 保存先のリポジトリ、ブランチ、パスを指定し、コミットメッセージを入力して「OK」をクリックします。

> ⚠️ **注意点**: この方法ではノートブック全体（出力結果を含む）が保存されます。競合（コンフリクト）を避けるため、コミット前に「出力をクリア」しておくことを推奨します（`編集` > `すべての出力をクリア`）。

### 2. GitHub上のノートブックをColabで開く
チームメンバーが作成したノートブックを閲覧・編集する場合、以下の方法で簡単にColabで開くことができます。

* **URLの書き換え**: GitHubのノートブックのURL `https://github.com/...` を `https://colab.research.google.com/github/...` に書き換えるだけで直接開けます。
* **「Colabで開く」バッジの設置**: ノートブックの先頭に以下のMarkdownを追加しておくと、クリックするだけで誰でも直接Colabで開けるようになります。

```markdown
[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/ユーザー名/リポジトリ名/blob/ブランチ名/パス/ノートブック名.ipynb)
```

### 4. コンフリクト（競合）を避けるためのチーム運用ルール
`.ipynb` ファイルはJSON形式で記述されており、差分（diff）が見づらく、コンフリクトが発生すると解消が非常に困難です。チームで開発する際は以下のルールを設けることを強くお勧めします。

1. **役割分担の明確化**: 同じノートブックを複数のメンバーが同時に編集しないようにタスクを切り分ける。
2. **コミット前の出力クリア**: メタデータや実行結果の画像などの差分を減らすため、保存・コミットする前に必ず `すべての出力をクリア` を実行する。
3. **Jupytextの導入（高度な開発）**: ノートブックを `.py` 形式（通常のPythonスクリプト）に自動変換してGit管理することで、Gitの差分を綺麗に保ち、コードレビューを容易にします。
